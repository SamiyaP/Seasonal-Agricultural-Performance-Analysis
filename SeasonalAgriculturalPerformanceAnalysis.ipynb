# Handling Missing Values

```python
#Handling missing values
#Here we will use median imputation for numerical columns

numeric_columns = df.select_dtypes(include=[np.number]).columns

for column in numeric_columns:
  if df[column].isnull().sum() > 0:
    print(f"Imputing missing values in column '{column}' with median")
    df[column].fillna(df[column].median(), inplace=True)
print("Remaining missing values : ",df.isnull().sum().sum())
```

# Irrigation Analysis

```python
#Irrigation Method vs Yield
irrigation_yield = (
    df.groupby('Irrigation_Method')['Yield_Tonnes_Ha']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))

sns.barplot(
    x=irrigation_yield.index,
    y=irrigation_yield.values
)

plt.title('Average Yield by Irrigation Method')
plt.xlabel('Irrigation Method')
plt.ylabel('Average Yield (Tonnes/Ha)')
plt.xticks(rotation=30)
plt.show()
```

Which irrigation method gives highest yield?
Drip

Key Insights :
Drip irrigation maximizes yield: By delivering water directly to plant roots, drip systems reduce waste and ensure crops get consistent moisture.

Sprinkler systems perform fairly well: Better than flood and rainfed, but less efficient than drip due to evaporation losses.

Flood irrigation underperforms: Despite heavy water use, yields are lower, showing inefficiency and potential soil damage.

Rainfed farming is least productive: While water-efficient, it depends entirely on rainfall, making yields less reliable.

```python
#Irrigation Method vs Water Efficiency
irrigation_efficiency = (
    df.groupby('Irrigation_Method')
    ['Water_Efficiency_t_per_1000m3']
    .mean()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10,5))

sns.barplot(
    x=irrigation_efficiency.index,
    y=irrigation_efficiency.values
)

plt.title('Water Efficiency by Irrigation Method')
plt.xlabel('Irrigation Method')
plt.ylabel('Water Efficiency (Tonnes/1000 m³)')
plt.xticks(rotation=30)
plt.show()
```

Rainfed method provides the best balance between productivity and water usage.

Ranking:

Rainfed → Highest efficiency (natural rainfall, no irrigation cost).

Drip → Next best, very efficient because water is delivered directly to plant roots.

Sprinkler → Moderate efficiency, but some water loss through evaporation.

Flood → Lowest efficiency, with large losses due to runoff and deep percolation.

# Soil Health Analysis

```python
#Soil pH vs Yield

plt.figure(figsize=(8,6))

sns.scatterplot(
    data=df,
    x='Soil_pH',
    y='Yield_Tonnes_Ha',
    alpha=0.6
)

plt.title('Soil pH vs Agricultural Yield')
plt.xlabel('Soil pH')
plt.ylabel('Yield (Tonnes/Ha)')
plt.show()
```

Key Insights :

Optimal zone: Yields are highest near neutral pH (6.5–7.0). This is the sweet spot where most crops thrive because nutrients are most available to plants.

Acidic soils (pH < 6): Yields drop sharply. Acidic conditions can lock up nutrients like phosphorus and increase toxic elements (e.g., aluminum).

Alkaline soils (pH > 7.5): Yields also decline. High alkalinity can reduce availability of micronutrients like iron and zinc.

Triangular pattern: Suggests that only within the optimal pH range do crops reach their maximum potential. Outside this range, yields taper off.
